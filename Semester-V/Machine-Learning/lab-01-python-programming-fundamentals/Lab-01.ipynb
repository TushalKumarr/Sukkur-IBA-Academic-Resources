## Exercise 01: Basic Math & Variable Swapping



```python
a, b = 15, 40

# swap values:
a, b = b, a

# calculations:
sum_result = a + b
product = a * b
division = a / b

# Display results
print(f"a = {a}")
print(f"b = {b}")
print(f"Sum = {sum_result}")
print(f"Product = {product}")
print(f"Division = {division:.2f}")
```

## Exercise 2: Username & Email Generator

```python
# store the name:
username = "Tushal Kumar"

# convert into lower case:
username = username.lower()

# replace space with underscore:
username = username.replace(" ", "_")

# creating domain email:
domain = "@iba-suk.edu.pk"
email = username + domain # string concatenation using +

# printing
print(f"Username: {username}")
print(f"Email: {email}")
```

## Exercise 3: String Reversal & Palindrome Checker
---

### What is a palindrome?

#### A palindrome is something that reads the same forward and backward.

#### Example: madam = madam



```python
text = "Nurses Run"

# removing space and converting into lower case:
cleaned = text.replace(" ", "").lower()

# reverse the string:
reversed_text = cleaned[::-1]

print(reversed_text)
```
