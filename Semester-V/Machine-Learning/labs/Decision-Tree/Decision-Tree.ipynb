# **Decison Trees**
---
A Decision Tree is a supervised machine learning algorithm used for both classification and regression.

It works like a flowchart, making decisions based on conditions.

For example, suppose we want to predict whether a student will crack an exam based on study hours and attendance.

[Decision Tree Documentation:](https://scikit-learn.org/1.6/modules/generated/sklearn.tree.DecisionTreeClassifier.html)

Decision-Tree-Diagram-Example-MindManager-Blog.png

````python
# Importing essential libraries:
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier
````


```
X = iris[["petal_length", "petal_width"]]
```

There are two separate things happening here.

Inner brackets: This creates a Python list containing two column names.

Outer brackets:This tells Pandas to select the columns specified in that list.

````python
# Loading "iris" dataset from the seaborn library:
iris = sns.load_dataset("iris")
iris.head()
````

````python
# Selecting input (X) and output (y) columns:
X_iris = iris[["petal_length", "petal_width"]]
y_iris = iris['species']
````

````python
# Create and train the model:
tree_clf = DecisionTreeClassifier(
    max_depth=2,
    random_state=42
)

tree_clf.fit(X_iris, y_iris)
````

**max_depth=2** limits the tree's depth, while

**random_state=42** helps make results reproducible.


**Synonyms**:
1.   Consistent
2.   Duplicable
3.   Repeatable



````python
# Make predictions:
prediction = tree_clf.predict([[5.0, 1.5]])

print(prediction)
````

````python
# Visualize the trained tree:
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plt.figure(figsize=(12, 7))

plot_tree(
    tree_clf,
    feature_names=X_iris.columns,
    class_names=tree_clf.classes_,
    filled=True,
    rounded=True
)


plt.show()
````
